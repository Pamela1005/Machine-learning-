```
import matplotlib.pyplot as plt
import pandas as pd

# Dataset
data = {
    'City': ['Delhi', 'Mumbai', 'Chennai', 'Kolkata'],
    '2015': [18, 20, 15, 13],
    '2017': [19, 22, 16, 14]
}

df = pd.DataFrame(data)

# Plotting
x = range(len(df['City']))
plt.figure(figsize=(8,5))

plt.bar(x, df['2015'], width=0.4, label='2015')
plt.bar([i + 0.4 for i in x], df['2017'], width=0.4, label='2017')

plt.xticks([i + 0.2 for i in x], df['City'])
plt.xlabel("Cities")
plt.ylabel("Population (in millions)")
plt.title("Population Comparison 2015 vs 2017")
plt.legend()
plt.show()
```

```
import pandas as pd

# Example dataframe like Superstore
df = pd.DataFrame({
    'Sales': [200, 1500, 300, 2500],
    'Profit': [-20, 300, -50, 500],
    'Quantity': [2, 55, 10, 70]
})

# Conditional formatting
def highlight_sales(val):
    if val > 1000:
        return "background-color: yellow"
    return ""

def profit_color(val):
    if val > 0:
        return "color: green"
    return "color: red"

def icon_quantity(val):
    if val > 50:
        return "background-color: lightgreen"
    return "background-color: lightcoral"

styled = (df.style
            .applymap(highlight_sales, subset=['Sales'])
            .applymap(profit_color, subset=['Profit'])
            .applymap(icon_quantity, subset=['Quantity'])
         )

styled.to_excel("formatted_superstore.xlsx", index=False)
print("File saved as formatted_superstore.xlsx")
```

```
import pandas as pd
import plotly.express as px

# Example city dataset
df = pd.DataFrame({
    'City': ['Delhi', 'Tokyo', 'London', 'Sydney'],
    'Latitude': [28.7041, 35.6762, 51.5074, -33.8688],
    'Longitude': [77.1025, 139.6503, -0.1278, 151.2093],
    'Population': [19000000, 37000000, 9000000, 5300000]
})

fig = px.scatter_geo(
    df,
    lat='Latitude',
    lon='Longitude',
    text='City',
    size='Population',
    title="World Cities Population Map",
    projection="natural earth"
)

fig.show()
```
